````python
"""
Normalize the formatting of RISC-V assembly test case blocks by removing internal blank lines, trimming trailing whitespace, and ensuring consistent separation between test cases while preserving a backup of the original file.
"""

from pathlib import Path
import shutil
import re

FILE_PATH = Path('TEXT/REPAIR_TEST_CASE_RAW.txt')
BACKUP_PATH = FILE_PATH.with_suffix(FILE_PATH.suffix + '.bak')

shutil.copy2(FILE_PATH, BACKUP_PATH)

text = FILE_PATH.read_text(encoding='utf-8').replace('\r\n', '\n').replace('\r', '\n')

pattern = re.compile(
    r'(#\s*\d+\s*# Extension Must Be Active:.*?)(?=\n\s*\n#\s*\d+\s*# Extension Must Be Active:|\Z)',
    re.DOTALL
)

blocks = []

for match in pattern.finditer(text):
    block = match.group(1).strip()

    lines = [
        line.rstrip()
        for line in block.splitlines()
        if line.strip()
    ]

    blocks.append('\n'.join(lines))

new_text = '\n\n'.join(blocks).strip() + '\n'

FILE_PATH.write_text(new_text, encoding='utf-8', newline='\n')

print(f'Done. Backup saved to: {BACKUP_PATH}')
````

````python
# 2 - UPDATE - 4_REPAIR_TEST_CASE_OUTPUT.txt - START

"""
Normalize repaired RISC-V assembly test cases by converting integer and floating-point register names into ABI aliases, replacing numeric CSR addresses with symbolic names, and simplifying trap-related expected results into direct mcause checks.
"""

# TIME START - BEGIN
import time
TIME_START = time.time()
# TIME START - FINISH

# IMPORT - BEGIN
from pathlib import Path
import re
# IMPORT - FINISH

# PATH - BEGIN
FILE_INPUT = Path('TEXT/4_REPAIR_TEST_CASE_OUTPUT.txt')
FILE_OUTPUT = Path('TEXT/4_REPAIR_TEST_CASE_OUTPUT_FINAL.txt')
FILE_REGISTER = Path('TEXT/REPAIR_TEST_CASE_REGISTER.txt')
# PATH - FINISH

# FUNCTION - BEGIN
def FUNCTION_TWOS_SIGNED(VALUE_INT: int, BIT_COUNT: int) -> int:
    VALUE_MASK = (1 << BIT_COUNT) - 1
    VALUE_INT &= VALUE_MASK
    VALUE_SIGN = 1 << (BIT_COUNT - 1)
    return VALUE_INT - (1 << BIT_COUNT) if (VALUE_INT & VALUE_SIGN) else VALUE_INT

def FUNCTION_CONVERT_HEX_IN_LINE(STRING_LINE: str) -> str:
    def FUNCTION_REPLACE(MATCH_VALUE: re.Match) -> str:
        STRING_HEX = MATCH_VALUE.group(1)
        IS_NEG = STRING_HEX.startswith('-')
        STRING_HEX_BODY = STRING_HEX[1:] if IS_NEG else STRING_HEX
        VALUE_INT = int(STRING_HEX_BODY, 16)
        if IS_NEG:
            VALUE_INT = -VALUE_INT
        if REGEX_RESULT_HEADER.match(STRING_LINE):
            return str(FUNCTION_TWOS_SIGNED(VALUE_INT, RESULT_BIT))
        return str(VALUE_INT)
    return REGEX_HEX.sub(FUNCTION_REPLACE, STRING_LINE)

def FUNCTION_REPLACE_REGISTER(STRING_LINE: str) -> str:
    STRING_LINE = REGEX_X.sub(lambda MATCH_VALUE: MAP_X_ABI.get(MATCH_VALUE.group(0), MATCH_VALUE.group(0)), STRING_LINE)
    STRING_LINE = REGEX_F.sub(lambda MATCH_VALUE: MAP_F_ABI.get(MATCH_VALUE.group(0), MATCH_VALUE.group(0)), STRING_LINE)
    return STRING_LINE

def FUNCTION_REPLACE_CSR_NUMBER(STRING_LINE: str) -> str:
    STRING_STRIPPED = STRING_LINE.lstrip()
    if not STRING_STRIPPED or STRING_STRIPPED.startswith('#'):
        return STRING_LINE
    MATCH_OP = re.match(r'^(\s*)([A-Za-z.][A-Za-z0-9_.]*)\s+(.*)$', STRING_LINE)
    if not MATCH_OP:
        return STRING_LINE
    STRING_INDENT = MATCH_OP.group(1)
    STRING_OP = MATCH_OP.group(2).lower()
    STRING_REST = MATCH_OP.group(3)
    if STRING_OP not in SET_CSR_OP:
        return STRING_LINE
    NL = '\n' if STRING_LINE.endswith('\n') else ''
    STRING_BEFORE_HASH, STRING_COMMENT_TEXT = (STRING_REST.split('#', 1) + [''])[:2]
    STRING_BEFORE_HASH = STRING_BEFORE_HASH.strip()
    STRING_COMMENT_TEXT = STRING_COMMENT_TEXT.strip()
    LIST_OPERAND = [x.strip() for x in STRING_BEFORE_HASH.split(',') if x.strip()]
    INDEX_CSR = None
    if STRING_OP in SET_CSR_PRR or STRING_OP in SET_CSR_PI:
        INDEX_CSR = 0 if len(LIST_OPERAND) >= 1 else None
    elif STRING_OP in SET_CSR_RRR or STRING_OP in SET_CSR_RRI:
        INDEX_CSR = 1 if len(LIST_OPERAND) >= 2 else None
    elif STRING_OP == 'csrr':
        INDEX_CSR = 1 if len(LIST_OPERAND) >= 2 else None
    if INDEX_CSR is not None:
        STRING_TOKEN = LIST_OPERAND[INDEX_CSR]
        if STRING_TOKEN.lower().startswith('0x') or re.fullmatch(r'\d+', STRING_TOKEN):
            VALUE_NUM = int(STRING_TOKEN, 16) if STRING_TOKEN.lower().startswith('0x') else int(STRING_TOKEN, 10)
            STRING_NAME = MAP_CSR_NUM_TO_NAME.get(VALUE_NUM)
            if STRING_NAME:
                LIST_OPERAND[INDEX_CSR] = STRING_NAME
    STRING_NEW = STRING_INDENT + STRING_OP + ' ' + ', '.join(LIST_OPERAND)
    if '#' in STRING_REST:
        if STRING_COMMENT_TEXT:
            return STRING_NEW + ' # ' + STRING_COMMENT_TEXT + NL
        return STRING_NEW + ' #' + NL
    return STRING_NEW + NL

def FUNCTION_COLLAPSE_MCAUSE_BLOCK(STRING_TEXT: str) -> str:
    LIST_PART = re.split(r'(?m)(?=^#\s*\d+\s*#\s*Extension Must Be Active:)', STRING_TEXT)
    LIST_OUTPUT = []
    REGEX_CSRR_MCAUSE = re.compile(r'^\s*csrr\s+([A-Za-z0-9_]+)\s*,\s*mcause\s*(?:#.*)?\s*$', re.I)
    REGEX_CSRRS_MCAUSE = re.compile(r'^\s*csrrs\s+([A-Za-z0-9_]+)\s*,\s*mcause\s*,\s*(x0|zero)\s*(?:#.*)?\s*$', re.I)
    REGEX_RESULT_REG = re.compile(r'^\s*#\s*Result:\s*([A-Za-z0-9_]+)\s*=\s*(.+?)\s*$', re.I)
    for EACH_PART in LIST_PART:
        if not EACH_PART.strip():
            LIST_OUTPUT.append(EACH_PART)
            continue
        LIST_LINE = EACH_PART.splitlines(True)
        REGISTER_DEST = None
        for EACH_INDEX in range(len(LIST_LINE) - 1, -1, -1):
            MATCH_MCAUSE = REGEX_CSRR_MCAUSE.match(LIST_LINE[EACH_INDEX]) or REGEX_CSRRS_MCAUSE.match(LIST_LINE[EACH_INDEX])
            if MATCH_MCAUSE:
                REGISTER_DEST = MATCH_MCAUSE.group(1)
                break
        if REGISTER_DEST is None:
            LIST_OUTPUT.append(EACH_PART)
            continue
        VALUE_REMOVED = None
        LIST_NEW_LINE = []
        for EACH_LINE in LIST_LINE:
            MATCH_RESULT = REGEX_RESULT_REG.match(EACH_LINE)
            if MATCH_RESULT and MATCH_RESULT.group(1) == REGISTER_DEST:
                if VALUE_REMOVED is None:
                    VALUE_REMOVED = MATCH_RESULT.group(2).strip()
                continue
            LIST_NEW_LINE.append(EACH_LINE)
        if VALUE_REMOVED is None:
            LIST_OUTPUT.append(EACH_PART)
            continue
        LIST_FILTERED = []
        REMOVED_CSRR = False
        for EACH_LINE in LIST_NEW_LINE:
            if (not REMOVED_CSRR) and (REGEX_CSRR_MCAUSE.match(EACH_LINE) or REGEX_CSRRS_MCAUSE.match(EACH_LINE)):
                REMOVED_CSRR = True
                continue
            LIST_FILTERED.append(EACH_LINE)
        STRING_INSERT = f'# Result: mcause = {VALUE_REMOVED}\n'
        VALUE_J = len(LIST_FILTERED)
        while VALUE_J > 0 and LIST_FILTERED[VALUE_J - 1].strip() == '':
            VALUE_J -= 1
        LIST_FILTERED.insert(VALUE_J, STRING_INSERT)
        LIST_OUTPUT.append(''.join(LIST_FILTERED))
    return ''.join(LIST_OUTPUT)
# FUNCTION - FINISH

# VARIABLE - BEGIN
RESULT_BIT = 32
MAP_X_ABI = {'x1': 'ra', 'x2': 'sp', 'x3': 'gp', 'x4': 'tp', 'x5': 't0', 'x6': 't1', 'x7': 't2', 'x8': 's0', 'x9': 's1', 'x10': 'a0', 'x11': 'a1', 'x12': 'a2', 'x13': 'a3', 'x14': 'a4', 'x15': 'a5', 'x16': 'a6', 'x17': 'a7', 'x18': 's2', 'x19': 's3', 'x20': 's4', 'x21': 's5', 'x22': 's6', 'x23': 's7', 'x24': 's8', 'x25': 's9', 'x26': 's10', 'x27': 's11', 'x28': 't3', 'x29': 't4', 'x30': 't5', 'x31': 't6'}
MAP_F_ABI = {'f0': 'ft0', 'f1': 'ft1', 'f2': 'ft2', 'f3': 'ft3', 'f4': 'ft4', 'f5': 'ft5', 'f6': 'ft6', 'f7': 'ft7', 'f8': 'fs0', 'f9': 'fs1', 'f10': 'fa0', 'f11': 'fa1', 'f12': 'fa2', 'f13': 'fa3', 'f14': 'fa4', 'f15': 'fa5', 'f16': 'fa6', 'f17': 'fa7', 'f18': 'fs2', 'f19': 'fs3', 'f20': 'fs4', 'f21': 'fs5', 'f22': 'fs6', 'f23': 'fs7', 'f24': 'fs8', 'f25': 'fs9', 'f26': 'fs10', 'f27': 'fs11', 'f28': 'ft8', 'f29': 'ft9', 'f30': 'ft10', 'f31': 'ft11'}
REGEX_X = re.compile(r'\bx(3[01]|[12]\d|[0-9])\b')
REGEX_F = re.compile(r'\bf(3[01]|[12]\d|[0-9])\b')
REGEX_HEX = re.compile(r'(?<![A-Za-z0-9_])(-?0x[0-9a-fA-F]+)(?![A-Za-z0-9_])')
REGEX_RESULT_HEADER = re.compile(r'^\s*#\s*Result\s*:', re.IGNORECASE)
SET_CSR_RRR = {'csrrw', 'csrrs', 'csrrc'}
SET_CSR_RRI = {'csrrwi', 'csrrsi', 'csrrci'}
SET_CSR_PRR = {'csrw', 'csrs', 'csrc'}
SET_CSR_PI = {'csrwi', 'csrsi', 'csrci'}
SET_CSR_PSEUDO = {'csrr'}
SET_CSR_OP = SET_CSR_RRR | SET_CSR_RRI | SET_CSR_PRR | SET_CSR_PI | SET_CSR_PSEUDO
MAP_CSR_NUM_TO_NAME = {}
REGISTER_CURRENT = None
STRING_TEXT = FILE_INPUT.read_text(encoding='utf-8', errors='replace').replace('\r\n', '\n')
LIST_LINE = STRING_TEXT.splitlines(True)
LIST_OUTPUT_LINE = []
# VARIABLE - FINISH

# CODE - BEGIN
for EACH_LINE in FILE_REGISTER.read_text(encoding='utf-8', errors='replace').splitlines():
    MATCH_REGISTER = re.match(r'^\s*REGISTER:\s*([A-Za-z0-9_]+)\s*$', EACH_LINE)
    if MATCH_REGISTER:
        REGISTER_CURRENT = MATCH_REGISTER.group(1)
        continue
    MATCH_NUMBER = re.match(r'^\s*NUMBER:\s*(0x[0-9a-fA-F]+|\d+)\s*$', EACH_LINE)
    if MATCH_NUMBER and REGISTER_CURRENT:
        STRING_NUMBER = MATCH_NUMBER.group(1)
        VALUE_NUMBER = int(STRING_NUMBER, 16) if STRING_NUMBER.lower().startswith('0x') else int(STRING_NUMBER, 10)
        MAP_CSR_NUM_TO_NAME.setdefault(VALUE_NUMBER, REGISTER_CURRENT)
        REGISTER_CURRENT = None

for EACH_LINE in LIST_LINE:
    EACH_LINE = FUNCTION_REPLACE_CSR_NUMBER(EACH_LINE)
    EACH_LINE = FUNCTION_REPLACE_REGISTER(EACH_LINE)
    LIST_OUTPUT_LINE.append(EACH_LINE)

STRING_FINAL = ''.join(LIST_OUTPUT_LINE)
STRING_FINAL = FUNCTION_COLLAPSE_MCAUSE_BLOCK(STRING_FINAL)
FILE_OUTPUT.write_text(STRING_FINAL, encoding='utf-8', newline='\n')
# CODE - FINISH

# TIME END - BEGIN
TIME_END = time.time()
DURATION_IN_SECOND = TIME_END - TIME_START
DAYS = int(DURATION_IN_SECOND // 86400)
DURATION_IN_SECOND %= 86400
HOURS = int(DURATION_IN_SECOND // 3600)
DURATION_IN_SECOND %= 3600
MINUTES = int(DURATION_IN_SECOND // 60)
DURATION_IN_SECOND %= 60
PARTS = []
if DAYS:
    PARTS.append(f'{DAYS} day(s)')
if HOURS:
    PARTS.append(f'{HOURS} hour(s)')
if MINUTES:
    PARTS.append(f'{MINUTES} minute(s)')
if DURATION_IN_SECOND or not PARTS:
    PARTS.append(f'{DURATION_IN_SECOND:.2f} second(s)')
print(' '.join(PARTS))
# TIME END - FINISH

# DONE - BEGIN
print('\n====\nDONE\n====')
# DONE - FINISH

# 2 - UPDATE - 4_REPAIR_TEST_CASE_OUTPUT.txt - END
````

````python
# 3 - CREATE - REPAIR_TEST_CASE_OUTPUT_EXTENSION.txt & 4_REPAIR_TEST_CASE_OUTPUT_FINAL_NON_XV6.txt - START

"""
Prepare finalized RISC-V assembly test cases for emulator execution by extracting extension-dependent test cases, adapting register and CSR bit-field checks for non-xv6 test harnesses, and distributing the test cases to QEMU, Spike, RVVM, and Renode.
"""

# TIME START - BEGIN
import time
TIME_START = time.time()
# TIME START - FINISH

# IMPORT - BEGIN
from pathlib import Path
import re
import shutil
# IMPORT - FINISH

# PATH - BEGIN
FILE_INPUT = Path('TEXT/4_REPAIR_TEST_CASE_OUTPUT_FINAL.txt')
FILE_OUTPUT = Path('TEXT/REPAIR_TEST_CASE_OUTPUT_EXTENSION.txt')
FILE_OUTPUT_NON_XV6 = Path('TEXT/4_REPAIR_TEST_CASE_OUTPUT_FINAL_NON_XV6.txt') # non xv6 special
# PATH - FINISH

# FUNCTION - BEGIN
def FUNCTION_FILTER_EXTENSION_BLOCK(LIST_LINE):
    LIST_OUTPUT = []
    LIST_CURRENT = []
    CURRENT_EXTENSION = None
    for EACH_LINE in LIST_LINE:
        MATCH = REGEX_EXTENSION_HEADER.match(EACH_LINE)
        if MATCH:
            if (
                LIST_CURRENT
                and CURRENT_EXTENSION
                and CURRENT_EXTENSION.lower() != 'none'
            ):
                while LIST_CURRENT and LIST_CURRENT[-1].strip() == '':
                    LIST_CURRENT.pop()
                if LIST_OUTPUT:
                    LIST_OUTPUT.append('\n')
                LIST_OUTPUT.extend(LIST_CURRENT)
            LIST_CURRENT = [EACH_LINE]
            CURRENT_EXTENSION = MATCH.group('ext')
        else:
            LIST_CURRENT.append(EACH_LINE)
    if (
        LIST_CURRENT
        and CURRENT_EXTENSION
        and CURRENT_EXTENSION.lower() != 'none'
    ):
        while LIST_CURRENT and LIST_CURRENT[-1].strip() == '':
            LIST_CURRENT.pop()
        if LIST_OUTPUT:
            LIST_OUTPUT.append('\n')
        LIST_OUTPUT.extend(LIST_CURRENT)
    return LIST_OUTPUT

def FUNCTION_CREATE_NON_XV6_TEXT(text: str) -> str:
    def FUNCTION_PARSE_INT(string_value: str):
        string_value = string_value.strip()
        try:
            return int(string_value, 0)
        except:
            return None

    MAP_GPR_ALIAS_TO_CANONICAL = {
        'zero': 'x0',  'x0': 'x0',
        'ra': 'x1',    'x1': 'x1',
        'sp': 'x2',    'x2': 'x2',
        'gp': 'x3',    'x3': 'x3',
        'tp': 'x4',    'x4': 'x4',
        't0': 'x5',    'x5': 'x5',
        't1': 'x6',    'x6': 'x6',
        't2': 'x7',    'x7': 'x7',
        's0': 'x8',    'fp': 'x8', 'x8': 'x8',
        's1': 'x9',    'x9': 'x9',
        'a0': 'x10',   'x10': 'x10',
        'a1': 'x11',   'x11': 'x11',
        'a2': 'x12',   'x12': 'x12',
        'a3': 'x13',   'x13': 'x13',
        'a4': 'x14',   'x14': 'x14',
        'a5': 'x15',   'x15': 'x15',
        'a6': 'x16',   'x16': 'x16',
        'a7': 'x17',   'x17': 'x17',
        's2': 'x18',   'x18': 'x18',
        's3': 'x19',   'x19': 'x19',
        's4': 'x20',   'x20': 'x20',
        's5': 'x21',   'x21': 'x21',
        's6': 'x22',   'x22': 'x22',
        's7': 'x23',   'x23': 'x23',
        's8': 'x24',   'x24': 'x24',
        's9': 'x25',   'x25': 'x25',
        's10': 'x26',  'x26': 'x26',
        's11': 'x27',  'x27': 'x27',
        't3': 'x28',   'x28': 'x28',
        't4': 'x29',   'x29': 'x29',
        't5': 'x30',   'x30': 'x30',
        't6': 'x31',   'x31': 'x31',
    }

    MAP_CANONICAL_TO_ABI = {
        'x5': 't0', 'x6': 't1', 'x7': 't2',
        'x28': 't3', 'x29': 't4', 'x30': 't5', 'x31': 't6',
        'x10': 'a0', 'x11': 'a1', 'x12': 'a2', 'x13': 'a3',
        'x14': 'a4', 'x15': 'a5', 'x16': 'a6', 'x17': 'a7',
        'x9': 's1', 'x18': 's2', 'x19': 's3', 'x20': 's4',
        'x21': 's5', 'x22': 's6', 'x23': 's7', 'x24': 's8',
        'x25': 's9', 'x26': 's10', 'x27': 's11', 'x8': 's0',
    }

    LIST_TEMP_POOL = [
        'x31', 'x29', 'x28', 'x7', 'x6', 'x5',      # t6, t4, t3, t2, t1, t0
        'x17', 'x16', 'x15', 'x14', 'x13', 'x12', 'x11', 'x10',   # a7..a0
        'x27', 'x26', 'x25', 'x24', 'x23', 'x22', 'x21', 'x20', 'x19', 'x18', 'x9', 'x8'  # s11..s0,s1
    ]

    def FUNCTION_CANONICALIZE_GPR(string_gpr: str) -> str:
        return MAP_GPR_ALIAS_TO_CANONICAL.get(string_gpr, string_gpr)

    def FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp):
        for each_temp_canonical in LIST_TEMP_POOL:
            if each_temp_canonical in set_used_gpr:
                continue
            if each_temp_canonical in set_reserved_temp:
                continue
            set_reserved_temp.add(each_temp_canonical)
            return MAP_CANONICAL_TO_ABI[each_temp_canonical]
        raise ValueError(
            f'No free temp register available. '
            f'used={sorted(set_used_gpr)} reserved={sorted(set_reserved_temp)}'
        )

    list_block = text.strip().split('\n\n')
    list_block_output = []

    for each_block in list_block:
        list_line = each_block.split('\n')
        list_line_output = []

        for each_line in list_line:
            stripped_line = each_line.strip()

            if stripped_line.startswith('# Result:'):
                string_prefix = '# Result:'
                string_body = stripped_line[len(string_prefix):].strip()
                if '|' in string_body:
                    string_separator = '|'
                elif '&' in string_body:
                    string_separator = '&'
                else:
                    string_separator = None
                
                if string_separator:
                    list_condition = [item.strip() for item in string_body.split(string_separator)]
                else:
                    list_condition = [string_body.strip()]
                list_condition_parsed = []
                set_used_gpr = set()

                for each_condition in list_condition:
                    match_value = re.match(
                        r'^([A-Za-z_][A-Za-z0-9_]*(?:\[\d+(?::\d+)?\])?)\s*(!=|=)\s*(.+)$',
                        each_condition
                    )
                    if not match_value:
                        list_condition_parsed.append({'kind': 'raw', 'text': each_condition})
                        continue

                    string_lhs = match_value.group(1).strip()
                    string_op = match_value.group(2).strip()
                    string_rhs = match_value.group(3).strip()

                    match_field = re.match(r'^([A-Za-z_][A-Za-z0-9_]*)\[(\d+)(?::(\d+))?\]$', string_lhs)
                    if match_field:
                        string_base_name = match_field.group(1)
                        int_msb = int(match_field.group(2))
                        int_lsb = int(match_field.group(3)) if match_field.group(3) is not None else int_msb
                    
                        if string_base_name in SET_NORMAL_REGISTERS:
                            set_used_gpr.add(FUNCTION_CANONICALIZE_GPR(string_base_name))
                            list_condition_parsed.append({
                                'kind': 'gpr_field',
                                'lhs': string_lhs,
                                'gpr': string_base_name,
                                'msb': int_msb,
                                'lsb': int_lsb,
                                'op': string_op,
                                'rhs': string_rhs,
                                'text': each_condition
                            })
                        else:
                            list_condition_parsed.append({
                                'kind': 'csr_field',
                                'lhs': string_lhs,
                                'csr': string_base_name,
                                'msb': int_msb,
                                'lsb': int_lsb,
                                'op': string_op,
                                'rhs': string_rhs,
                                'text': each_condition
                            })

                flag_need_edit = any(each_item['kind'] in {'csr', 'csr_field', 'gpr_field'} for each_item in list_condition_parsed)
                if not flag_need_edit:
                    list_line_output.append(each_line)
                    continue

                set_reserved_temp = set()
                map_csr_to_temp = {}
                list_extra_instruction = []
                list_condition_output = []

                for each_item in list_condition_parsed:
                    if each_item['kind'] == 'raw':
                        list_condition_output.append(each_item['text'])
                        continue

                    if each_item['kind'] == 'gpr':
                        list_condition_output.append(each_item['text'])
                        continue

                    if each_item['kind'] == 'csr':
                        string_csr_name = each_item['csr']
                        if string_csr_name == 'mcause':
                            list_condition_output.append(f't5 {each_item["op"]} {each_item["rhs"]}')
                            continue

                        if string_csr_name not in map_csr_to_temp:
                            map_csr_to_temp[string_csr_name] = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                            list_extra_instruction.append(f'csrr {map_csr_to_temp[string_csr_name]}, {string_csr_name}')

                        list_condition_output.append(f'{map_csr_to_temp[string_csr_name]} {each_item["op"]} {each_item["rhs"]}')
                        continue

                    if each_item['kind'] == 'gpr_field':
                        string_gpr_name = each_item['gpr']
                        int_msb = each_item['msb']
                        int_lsb = each_item['lsb']
                        int_width = int_msb - int_lsb + 1
                        int_mask = (1 << int_width) - 1

                        string_temp_value = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                        list_extra_instruction.append(f'mv {string_temp_value}, {string_gpr_name}')

                        if int_lsb != 0:
                            list_extra_instruction.append(f'srli {string_temp_value}, {string_temp_value}, {int_lsb}')

                        if int_width == 1:
                            list_extra_instruction.append(f'andi {string_temp_value}, {string_temp_value}, 0x1')
                        else:
                            if int_mask <= 0xfff:
                                list_extra_instruction.append(f'andi {string_temp_value}, {string_temp_value}, {hex(int_mask)}')
                            else:
                                string_temp_mask = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                                list_extra_instruction.append(f'li {string_temp_mask}, {hex(int_mask)}')
                                list_extra_instruction.append(f'and {string_temp_value}, {string_temp_value}, {string_temp_mask}')

                        list_condition_output.append(f'{string_temp_value} {each_item["op"]} {each_item["rhs"]}')
                        continue

                    if each_item['kind'] == 'csr_field':
                        string_csr_name = each_item['csr']
                        int_msb = each_item['msb']
                        int_lsb = each_item['lsb']
                        int_width = int_msb - int_lsb + 1
                        int_mask = (1 << int_width) - 1

                        if int_width == 1 and int_msb == 63 and int_lsb == 63:
                            string_temp_mask = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                            string_temp_value = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                            list_extra_instruction.append(f'li {string_temp_mask}, 1 << 63')
                            list_extra_instruction.append(f'csrr {string_temp_value}, {string_csr_name}')
                            list_extra_instruction.append(f'and {string_temp_mask}, {string_temp_mask}, {string_temp_value}')

                            int_rhs = FUNCTION_PARSE_INT(each_item['rhs'])
                            if int_rhs is not None:
                                string_rhs_output = hex((1 << 63) if int_rhs != 0 else 0x0)
                            else:
                                string_rhs_output = each_item['rhs']

                            list_condition_output.append(f'{string_temp_mask} {each_item["op"]} {string_rhs_output}')
                            continue

                        string_temp_value = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                        list_extra_instruction.append(f'csrr {string_temp_value}, {string_csr_name}')

                        if int_lsb != 0:
                            list_extra_instruction.append(f'srli {string_temp_value}, {string_temp_value}, {int_lsb}')

                        if int_width == 1:
                            list_extra_instruction.append(f'andi {string_temp_value}, {string_temp_value}, 0x1')
                        else:
                            if int_mask <= 0xfff:
                                list_extra_instruction.append(f'andi {string_temp_value}, {string_temp_value}, {hex(int_mask)}')
                            else:
                                string_temp_mask = FUNCTION_ALLOCATE_TEMP(set_used_gpr, set_reserved_temp)
                                list_extra_instruction.append(f'li {string_temp_mask}, {hex(int_mask)}')
                                list_extra_instruction.append(f'and {string_temp_value}, {string_temp_value}, {string_temp_mask}')

                        list_condition_output.append(f'{string_temp_value} {each_item["op"]} {each_item["rhs"]}')
                        continue

                for each_instruction in list_extra_instruction:
                    list_line_output.append(each_instruction)

                string_join = f' {string_separator} ' if string_separator else ''
                list_line_output.append(f'{string_prefix} ' + (string_join.join(list_condition_output) if string_separator else list_condition_output[0]))
                continue

            list_line_output.append(each_line)

        list_block_output.append('\n'.join(list_line_output))

    return '\n\n'.join(list_block_output).strip()
# FUNCTION - FINISH

# VARIABLE - BEGIN
REGEX_EXTENSION_HEADER = re.compile(
    r'^\s*#\s*\d+\s*#\s*Extension Must Be Active:\s*(?P<ext>\S+)',
    re.IGNORECASE
)

REGEX_RESULT_EQUAL = re.compile(
    r'^(#\s*Result:\s*)([A-Za-z_][A-Za-z0-9_]*)\s*=\s*(.+)$'
)

SET_NORMAL_REGISTERS = {
    'zero', 'ra', 'sp', 'gp', 'tp',
    't0', 't1', 't2', 't3', 't4', 't5', 't6',
    's0', 's1', 's2', 's3', 's4', 's5', 's6', 's7', 's8', 's9', 's10', 's11',
    'a0', 'a1', 'a2', 'a3', 'a4', 'a5', 'a6', 'a7',
    'fp',
    'x0', 'x1', 'x2', 'x3', 'x4', 'x5', 'x6', 'x7', 'x8', 'x9', 'x10', 'x11', 'x12', 'x13', 'x14', 'x15',
    'x16', 'x17', 'x18', 'x19', 'x20', 'x21', 'x22', 'x23', 'x24', 'x25', 'x26', 'x27', 'x28', 'x29', 'x30', 'x31',
    'ft0', 'ft1', 'ft2', 'ft3', 'ft4', 'ft5', 'ft6', 'ft7', 'ft8', 'ft9', 'ft10', 'ft11',
    'fs0', 'fs1', 'fs2', 'fs3', 'fs4', 'fs5', 'fs6', 'fs7', 'fs8', 'fs9', 'fs10', 'fs11',
    'fa0', 'fa1', 'fa2', 'fa3', 'fa4', 'fa5', 'fa6', 'fa7',
    'f0', 'f1', 'f2', 'f3', 'f4', 'f5', 'f6', 'f7', 'f8', 'f9', 'f10', 'f11', 'f12', 'f13', 'f14', 'f15',
    'f16', 'f17', 'f18', 'f19', 'f20', 'f21', 'f22', 'f23', 'f24', 'f25', 'f26', 'f27', 'f28', 'f29', 'f30', 'f31',
    'v0', 'v1', 'v2', 'v3', 'v4', 'v5', 'v6', 'v7', 'v8', 'v9', 'v10', 'v11', 'v12', 'v13', 'v14', 'v15',
    'v16', 'v17', 'v18', 'v19', 'v20', 'v21', 'v22', 'v23', 'v24', 'v25', 'v26', 'v27', 'v28', 'v29', 'v30', 'v31',
}
# VARIABLE - FINISH

# CODE - BEGIN
with open(FILE_INPUT, 'r', encoding='utf-8') as f:
    LIST_LINE = f.readlines()

LIST_OUTPUT = FUNCTION_FILTER_EXTENSION_BLOCK(LIST_LINE)
TEXT_OUTPUT = ''.join(LIST_OUTPUT).rstrip('\n').rstrip()

with open(FILE_OUTPUT, 'w', encoding='utf-8', newline='\n') as f:
    f.write(TEXT_OUTPUT)

TEXT_INPUT = ''.join(LIST_LINE).replace('\r\n', '\n').strip()
TEXT_OUTPUT_NON_XV6 = FUNCTION_CREATE_NON_XV6_TEXT(TEXT_INPUT)

with open(FILE_OUTPUT_NON_XV6, 'w', encoding='utf-8', newline='\n') as f:
    f.write(TEXT_OUTPUT_NON_XV6)
# CODE - FINISH

# TIME END - BEGIN
TIME_END = time.time()
DURATION_IN_SECOND = TIME_END - TIME_START
DAYS = int(DURATION_IN_SECOND // 86400)
DURATION_IN_SECOND %= 86400
HOURS = int(DURATION_IN_SECOND // 3600)
DURATION_IN_SECOND %= 3600
MINUTES = int(DURATION_IN_SECOND // 60)
DURATION_IN_SECOND %= 60
PARTS = []
if DAYS:
    PARTS.append(f'{DAYS} day(s)')
if HOURS:
    PARTS.append(f'{HOURS} hour(s)')
if MINUTES:
    PARTS.append(f'{MINUTES} minute(s)')
if DURATION_IN_SECOND or not PARTS:
    PARTS.append(f'{DURATION_IN_SECOND:.2f} second(s)')
print(' '.join(PARTS))
# TIME END - FINISH

# DONE - BEGIN
print('\n====\nDONE\n====')
# DONE - FINISH

# BONUS - BEGIN
SRC = FILE_INPUT
DST = Path('../8_EMULATOR/1_QEMU_XV6/TEXT/1_QEMU_XV6_INPUT.txt')
shutil.copy(SRC, DST)
SRC = FILE_INPUT
DST = Path('../8_EMULATOR/2_QEMU/TEXT/1_QEMU_INPUT.txt')
shutil.copy(SRC, DST)
DST = Path('../8_EMULATOR/3_SPIKE/TEXT/1_SPIKE_INPUT.txt')
shutil.copy(SRC, DST)
DST = Path('../8_EMULATOR/4_RVVM/TEXT/1_RVVM_INPUT.txt')
shutil.copy(SRC, DST)
DST = Path('../8_EMULATOR/5_RENODE/TEXT/1_RENODE_INPUT.txt')
_ = shutil.copy(SRC, DST)
# BONUS - FINISH

# 3 - CREATE - REPAIR_TEST_CASE_OUTPUT_EXTENSION.txt & 4_REPAIR_TEST_CASE_OUTPUT_FINAL_NON_XV6.txt - END
````
