```python
from pathlib import Path
import re

INPUT_PATTERN = "result_*.txt"


def clean_test_cases(text: str) -> str:
    text = text.replace("\r\n", "\n").replace("\r", "\n").strip()

    text = re.sub(
        r"(?im)^\s*#+\s*ASSEMBLY\s*#*\s*$",
        "##### ASSEMBLY #####",
        text,
    )

    lines = text.splitlines()
    blocks = []
    current_block = []

    for raw_line in lines:
        line = raw_line.strip()

        if not line:
            continue

        if line.startswith("# Test Case Rule:"):
            if current_block:
                blocks.append(current_block)
            current_block = [line]
            continue

        if current_block:
            current_block.append(line)

    if current_block:
        blocks.append(current_block)

    cleaned = "\n\n".join("\n".join(block) for block in blocks)

    if cleaned:
        cleaned += "\n"

    return cleaned


input_files = sorted(Path(".").glob(INPUT_PATTERN))

for input_path in input_files:
    output_path = input_path.with_name(f"clean_{input_path.name}")

    text = input_path.read_text(encoding="utf-8")
    cleaned = clean_test_cases(text)
    output_path.write_text(cleaned, encoding="utf-8")

    print(f"wrote: {output_path}")
```
